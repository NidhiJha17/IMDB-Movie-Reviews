```python
import matplotlib.pyplot as plt

labels = data["sentiment"]

labels.value_counts().plot(
    kind="bar",
    color=["steelblue", "tomato"],
    edgecolor="white"
)

plt.title("Class Distribution")
plt.xlabel("Label")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
